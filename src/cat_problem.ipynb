```python
#Libraries
import numpy as np
import pandas as pd
import statsmodels.api as sm
```

```python
#Loading the data
train = pd.read_csv(r'D:\Study folder\ASL Code\Exercise 2\data\cats_train.csv')
test = pd.read_csv(r'D:\Study folder\ASL Code\Exercise 2\data\cats_test.csv')
```

```python
#Building X Matrix and y vector for training data
X_train = np.column_stack([
    np.ones(len(train)),
    train['Sex'].values,
    train['Body_Weight'].values
])

y_train = train['Heart_Weight'].values
```

```python
# Compute Theta
theta_hat = np.linalg.inv(X_train.T @ X_train) @ (X_train.T @ y_train)
theta_hat
```

```python
#Using built in regression
X_train_sm = sm.add_constant(train[["Sex", "Body_Weight"]])
model = sm.OLS(y_train, X_train_sm).fit()
model.summary()
```

```python
#part c training error 
y_pred_train = X_train @ theta_hat
train_error = np.mean((y_train - y_pred_train)**2)
train_error
```

```python
#test error 
X_test = np.column_stack([
    np.ones(len(test)),
    test["Sex"].values,
    test["Body_Weight"].values
])

y_test = test["Heart_Weight"].values
y_pred_test = X_test @ theta_hat

test_error = np.mean((y_test - y_pred_test)**2)
test_error
```
